3.  Write a media query that changes the background color of the body to light
grey when the screen width is less than 600px.
    - Media query is a css technique introduced in CSS3 that allows you to apply different style dependent on characteristics of user's devices, Such as screen size, resolution, orientation, etc.
     It's specially useful for responsive design, which help to make website good look on devices of all sizes - from desktops to smartphones

      - The Media Query Code :
        
        ```CSS

            @media (max-width: 599){
              body{
                background-color: lightgrey;
              }
            }
        ```

      - Full Example of Webpage :
        
        ```HTML

          <!DOCTYPE html>
          <html>
            <head>
              <meta charset="UTF-8">
              <title> Media Query Example </title>
              <link rel="stylesheet" href="styles.css" />
            </head>
            <body>
              <p>hello my name rutik</p>
            </body>
          </html>

        ```
        ```styles.css

            body {
            background-color: white;
            }  

            @media (max-width: 599){
             body{
            background-color: lightgrey;            
             }
            }

         ```

    - **SUMMARY**   
        * A media query let you apply based on screen condition
        * @media ( max-width: 599px ) target devices smaller than 600px
        * Uses to change style (like ```background-color```)for mobile responsiveness

7. Explain the difference between position: absolute; and position: fixed;
in CSS.
  - In CSS both ```Position:absolute``` and ```Position:relative``` are used to remove element from the normal document flow and position them precisely using the ```top```,```right```,```bottom```, and ```left``` properties.
  * ```Position: absolute```


    *  ** Positioning Context:** An element with position: absolute; is positioned relative to its nearest positioned ancestor (an ancestor with position set to relative, absolute, or fixed). If no such ancestor exists, it’s positioned relative to the initial containing block (usually the ```<html>``` or ```<body>```).
    *  **Scrolling Behavior:** The element moves with the page when the user scrolls.
    *   **Use Case:** When you want to place an element inside a container or section and have it stay positioned within that container.

    **CSS**
    ```CSS
    .parent{
      position: relative;
    }

    .child{
      position: absolute;
      top: 10px;
      left: 20px;
    }
    ```
    

    **```Position: fixed```**


    *   **Positioning Context:** An element with position: fixed; is positioned relative to the browser window (viewport), regardless of any ancestor.
    *   **Scrolling Behavior:** The element does not move when the page is scrolled — it stays "fixed" in place.
    *  **Use Case:** For elements that should always be visible on the screen, like sticky headers, footers, or floating buttons.

    **CSS**

    ```CSS

    .fixed-header{
      position: fixed;
      top: 0;
      left: 0;
      width: 100%;
    }









      


   

8. Create a layered layout where a semi-transparent overlay covers an image
using z-index. Provide the HTML and CSS code snippets.
  -
  **HTML**
  ```HTML

  <!DOCTYPE html>
  <html>
    <head>
      <meta charset="UTF-8">
      <title>webpage of z-index</title>
      <link rel="stylesheet" href="styles.css" />
    </head>
    <body>

      <div class="image-container">
       <img src="your-image.jpg" alt="Background" class="background-image">
       <div class="overlay"></div>
      </div>

    </body>
  </html>
   ```

   **CSS (styles.css):**

   ```CSS
   .image-container{
    position: relative;
    width: 600px;
    height: 400px;
    overflow: hidden;
   }  

   .background-image{
    height: 100%;
    width: 100%;
    object-fit: cover;
    display: block;
    position: relative;
    z-index: 1;
   }

   .overlay{
    position: absolute;
    top: 0;
    left: 0;
    width: 100%;
    height: 100%;
    background-color: rgba(0,0,0,0.5);
    z-index: 2;
   }
   ```
    
  -  Key concepts :
      * ```position: relative;``` on the container creates a positioning context.
      * ```z-index: 1``` for the image ensures it's behind the overlay.
      * ```z-index: 2``` on the overlay puts it above the image.
      * ```rgba(0, 0, 0, 0.5)``` gives the overlay a 50% black transparency.
